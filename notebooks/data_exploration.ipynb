```python

import pandas as pd
from pathlib import Path

# Notebook is inside notebooks/, so go up one directory
data_dir = Path("../data")

train_df = pd.read_csv(data_dir / "healthver_train.csv")
dev_df = pd.read_csv(data_dir / "healthver_dev.csv")
test_df = pd.read_csv(data_dir / "healthver_test.csv")

print("Training samples:", len(train_df))
print("Validation samples:", len(dev_df))
print("Testing samples:", len(test_df))

print("\nColumns:", train_df.columns.tolist())

display(train_df.head())


print(train_df["label"].value_counts())

print(train_df[["claim", "evidence", "label"]].isnull().sum())

print("Unique claims:", train_df["claim"].nunique())
print("Unique evidence:", train_df["evidence"].nunique())

```
